**Also mention any two industries where Tableau is commonly used and explain one use case.**

**Ans-**

Tableau is a data visualization and business intelligence tool that helps turn raw data into interactive dashboards and insights. It is widely used in industries like finance and healthcare. For example, in banking, Tableau dashboards are used to monitor transactions and detect fraud patterns quickly, helping teams act in real time.

**Q4. What insight can be gained by comparing Sales and Profit together?**

**Then explain how you would create a chart to analyze Profit by Sub-Category.**

**Ans-**

Comparing Sales and Profit together gives deeper insight than looking at sales alone. **Q8. What is the difference between a Worksheet and a Dashboard in Tableau?**

**Why is a dashboard more useful for decision-makers?**

**Ans-**

**Difference between Worksheet and Dashboard in Tableau:**

*   A Worksheet is a single view where you build one chart or visualization (e.g., bar chart, line chart, map).
*   A Dashboard is a collection of multiple worksheets and other elements (like filters, text, images) combined into one interactive view.

**Why a Dashboard is more useful for decision‑makers:**

*   Dashboards provide a consolidated view of key metrics in one place.
*   They allow interactive filtering and comparisons across multiple charts.
*   Decision‑makers can quickly spot trends, correlations, and problem areas without switching between separate worksheets.

**Ans-**

Tableau Prep Tool for cleaning and preparing data before analysis. Real‑world data often has errors (nulls, duplicates, wrong types). Prep ensures clean, reliable data for accurate insights.

**Remove Nulls** → In profile pane, filter field → Exclude Null Values.

**Remove Duplicates** → Use Remove Duplicates option based on unique ID.
